الجزء 3 — استخراج البيانات: SELECT, WHERE, ORDER BY 

```python
# عرض  كل المنتجات 
cursor.execute ('''
SELECT * from products
''')
result = cursor.fetchall()
print(result)
```

```python
#عرض اسم وسعر المنتجات يلي سعرها أكبر من 50
cursor.execute('''
SELECT product_name,
        price
FROM products
WHERE price > 50

''')
result = cursor.fetchall()
print(result)
```

```python
#عرض كل عمليات البيع بمدينة معينة (مثلاً "عمان")، مرتبة حسب التاريخ تصاعديً
cursor.execute('''
SELECT * FROM sales
WHERE city = "Amman"
ORDER BY sale_date Desc
''')
result = cursor.fetchall()
print(result)
```

```python
#عرض المنتجات يلي فئتها "electronics" و سعرها أقل من 100 
cursor.execute('''
SELECT product_name
FROM products
WHERE category = "Electronics" AND price <200
''')
result = cursor.fetchall()
print(result)
```

```python
#عرض أعلى 5 عمليات بيع من ناحية الكمية
cursor.execute('''
SELECT * FROM sales
ORDER BY quantity DESC LIMIT 5
''')
result = cursor.fetchall()
print(result)
```

الجزء 4 — التلخيص: GROUP BY ودوال التجميع

```python
#اجمالي الكمية المباعة لكل منتج
cursor.execute('''
SELECT product_id,
        SUM(quantity)
FROM sales
GROUP BY product_id
''')
result = cursor.fetchall()
print(result)
```

```python
#عدد عمليات البيع بكل مدينة 
cursor.execute('''
SELECT city,
        COUNT(sales_id)
FROM sales
GROUP BY city
 ''')
result = cursor.fetchall()
print(result)
```

```python
# متوسط سعر المنتجات بكل فئة
cursor.execute('''
SELECT AVG(price),
        category
FROM products
GROUP BY category
''')
result = cursor.fetchall()
print(result)
```

```python
#عدد عمليات البيع بكل مدينة 
cursor.execute('''
SELECT city,
        COUNT(sales_id)"NUM OF ORDERS"
FROM sales
GROUP BY city
ORDER BY "NUM OF ORDERS" DESC LIMIT 1
 ''')
result = cursor.fetchall()
print(result)
```

الجزء 5 — المشروع المصغّر: 5 أسئلة بزنس حقيقية

```python
#شو أكثر 3 منتجات مبيعًا (من ناحية الكمية الإجمالية)؟
cursor.execute(''' 
SELECT  products.product_name,
        SUM(sales.quantity)"Total Quantity"
FROM products
JOIN sales
ON products.product_id = sales.product_id
GROUP BY products.product_name
ORDER BY "Total Quantity" DESC LIMIT 3
''')
result = cursor.fetchall()
print(result)
```

```python
#شو إجمالي الإيراد (الكمية × السعر) لكل منتج؟ 
cursor.execute('''
SELECT  products.product_name,
        SUM((sales.quantity)*(products.price))"Total Revenue"
FROM products
LEFT JOIN sales
ON products.product_id = sales.product_id
GROUP BY product_name
ORDER BY "Total Revenue" DESC 
''')
result = cursor.fetchall()
print(result)
```

```python
#شو المدينة الأعلى بإجمالي الكمية المباعة
cursor.execute(''' 
SELECT  sales.city,
        SUM(sales.quantity)"Total Quantity"
FROM sales
GROUP BY city
ORDER BY "Total Quantity" DESC LIMIT 1
''')
result = cursor.fetchall()
print(result)
```

```python
#شو الفئة (category) الأكثر تحقيقًا للمبيعات من ناحية عدد العمليات
cursor.execute('''
SELECT products.category,
        COUNT(sales.sales_id)"Total Orders"
FROM products
JOIN sales
ON products.product_id = sales.product_id
GROUP BY category
ORDER BY "Total Orders" DESC LIMIT 1
''')
result = cursor.fetchall()
print(result)
```

```python
#شو متوسط الكمية المباعة بكل عملية بيع بشكل عام (على كل الجدول، بدون GROUP BY)؟
cursor.execute('''
SELECT AVG(quantity)"Avg. Quantuty of Sales"
FROM sales
''')
result = cursor.fetchall()
print(result)
```

أظهرت البيانات أن بعض المنتجات متل الهواتف الذكية تحقق حجم مبيعات أعلى من غيرها من حيث الكمية والإيرادات، الذي يساعد المتجر على معرفة المنتجات الأكثر طلب. كما تبين أن مدينة عمان تتصدر من ناحية إجمالي الكميات المباعة وعدد عمليات البيع، مما قد يشير إلى قوة الطلب فيها. الفئة الأكثر مبيعًا الإلكترونيات : يمكن التركيز عليها عند التخطيط للمخزون والعروض القادمة.و بشكل عام، متوسط كمية البيع يعطي فكرة عن حجم الطلب في كل عملية شراء.
