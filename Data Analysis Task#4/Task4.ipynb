```python
import numpy as np
import pandas as pd
import seaborn as sns
titanic = sns.load_dataset('titanic')
titanic.head()
```

الجزء الأول اكتشاف و معالجة القيم المفقودة

```python
print("Missing Data Count",titanic.isnull().sum().sort_values(ascending = False))
```

```python
age_filled = titanic["age"].fillna(titanic["age"].mean())
titanic["age_filled"]=age_filled
titanic.isnull().sum().sort_values(ascending = False)
```

```python
#بالنسبة لعمود deck فسبة القيم المفقودة ما يقارب ال 77% بما معناه ان معظم القيم مفقودة لذلك فانه من الافضل كمحلل بيانات القيام بحذف العمود كاملا
# لانه وجود عمود مفقود القيم بشكل كبير مع عدم امكانية توفر تلك القيم فمن الممكن ان يقود ذلك الى نتائج مضلله
```

```python
titanic.head()
titanic["embarked"].mode()
titanic["embarked"] = titanic["embarked"].fillna(titanic["embarked"].mode()[0]) 
titanic.isnull().sum().sort_values(ascending = False)
```

الجزء الثالث : تلخيص البيانات 

```python
avg_fare_by_pclass = titanic.groupby("pclass")["fare"].mean().sort_values(ascending = False)
max_avg_fare = avg_fare_by_pclass.max()
max_pclass_avg_fare = avg_fare_by_pclass.idxmax()
print("متوسط سعر التذكرة لكل فئة ركاب")
print(avg_fare_by_pclass.round(2))
print("اعلى فئة بناءا علىمتوسط سعر التذكره : ",max_pclass_avg_fare)
print("أعلى متوسط سعر تذكرة : ",round(max_avg_fare,2))
```

```python
#نسب النجاة بناءا على الجنس 
survive_percentage_by_sex = titanic.groupby("sex")["survived"].mean()*100
print("نسب النجاة بناءا على الجنس")
print(survive_percentage_by_sex.round(2))
print("أعلى فئة نسبة نجاة هي : ",survive_percentage_by_sex.round(2).idxmax(),"بنسبة نجاة تساوي : ",survive_percentage_by_sex.round(2).max())
print("أقل فئة نسبة نجاة هي : ",survive_percentage_by_sex.round(2).idxmin(),"بنسبة نجاة تساوي : ",survive_percentage_by_sex.round(2).min())
```

```python
#نسب النجاة بناءا على الجنس و الفئة
survive_percentage_by_sex_pclass = (titanic.groupby(["sex", "pclass"])["survived"].mean().sort_values(ascending=False) * 100)

#النسبة الاعلى حسب الفئة و الجنس
highest_survive_percentage = survive_percentage_by_sex_pclass.max()
highest_group = survive_percentage_by_sex_pclass.idxmax()

#النسبة الاقل حسب الفئة و الجنس
lowest_survive_percentage = survive_percentage_by_sex_pclass.min()
lowest_group = survive_percentage_by_sex_pclass.idxmin()

print("نسب النجاة حسب الجنس والفئة : ")
print(survive_percentage_by_sex_pclass.round(2))

print("\nأعلى نسبة نجاة:")
print("الجنس : ", highest_group[0])
print("الفئة : ", highest_group[1])
print("النسبة : ", round(highest_survive_percentage, 2))

print("\nأقل نسبة نجاة:")
print("الجنس : ", lowest_group[0])
print("الفئة : ", lowest_group[1])
print("النسبة : ", round(lowest_survive_percentage, 2))
```

الجزء الرابع : دمج الجداول

```python
customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'name': ['أحمد', 'سارة', 'محمد', 'ليان'],
    'city': ['عمان', 'إربد', 'الزرقاء', 'عمان']})

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105],
    'customer_id': [1, 2, 1, 3, 5],
    'amount': [150, 200, 90, 300, 60]})
```

```python
inner_merged_data = pd.merge(customers,orders,on = "customer_id",how = "inner")
print(inner_merged_data)
# inner merge : اخد الريكورد يلي المفتاح تبعه موجود بكلا ال data frames
# customer id = 1 كان موجود بالجدولين لكن كان اله طلبين بالتالي اخدهم 
# customer_id = 2,3 كان موجود بالجدولين وله طلب واحد فاخده
# customer_id = 4,5 تواجدو بجدول و لم يتواجدو بالتاني لذلك تم استثناؤهم
```

```python
left_merged_data = pd.merge(customers,orders,on = "customer_id",how = "left")
print(left_merged_data)
# left merge : اخد كل الداتا الموجوده بالجدول الاول و اخد الداتا يلي تطابقت مع الجدول التاني
# customer_id = 1,2,3,4  هدول موجودين بالجدول الايسر فاخد كل صفوفهم و طابقهم بالجدول التاني يلي هو الايمن
# customer_id = 5 هاد غير متواجد بالايسر ف ما اخده من الاساس
```

```python
outer_merged_data = pd.merge(customers,orders,on = "customer_id",how = "outer")
print(outer_merged_data)
# هون اخد الداتا كلها يلي بالجدول الاول و التاني
# واي نقص بالبيانات كان يتركه Nan
```

left merge : بياخد كل شيء موجود بالجدول الاول و اي شيء بطابقه بالجدول التاني( كل شيء من الجدول الاول والبيانات المطابقة من التاني)
inner merge : بياخد المشترك بين الجدولين (المشترك فقط)
outer merge : بياخد الداتا كاملة لي بالجدولين و اي نقص بالبيانات بتركه فارغ (كل شيء من الجدولين)

```python
orders_feb = pd.DataFrame({'order_id': [105, 106, 107, 108, 109],
    'customer_id': [5, 2, 4, 3, 5],
    'amount': [60, 320, 140, 460, 78]})
total_orders = pd.concat([orders,orders_feb])
print(total_orders)
```

الجزء الخامس : المشروع التكاملي

```python
customers = pd.DataFrame({
    "Customer Name": ["Ali","Omer","Rasha","Mohammad","Basel","Rasha","Amal","Omer","Laith","Waleed","Waleed"],
    "customer_id": [1,2,3,4,5,3,6,2,7,8,8],
    "City": ["Amman","Zarqaa","Karak","Salt","Irbid","Karak",np.nan,"Zarqaa",np.nan,"Amman","Amman"]})
orders = pd.DataFrame({ 
    'order_id': [100, 101, 102, 103, 104,105,106,107,108,109,110,111,112,113,114], 
    'customer_id': [1, 2, 1, 3, 5,6,7,2,2,3,5,5,7,8,8], 
    'amount': [150, 200, 90, 300, 60,700,780,190,178,650,136,198,445,187,65]})

```

```python
#تنظيف الجدول الاول من التكرارات و القيم المفقودة
print(customers.isnull().sum()) # يوجد قيمتين مفقودتين في city
customers = customers.fillna("Not Available") # تم تعبئة القيم المفقودة ب not available
print(customers)
print(customers.duplicated().sum()) # يوجد 3 تكرارات
customers.drop_duplicates(keep = 'first',inplace = True) # تم حذف التكرار و حفظ النتيجة الاولى 
print(customers)
```

```python
#تنظيف الجدول الثاني من التكرارات و القيم المفقودة
print(orders.isnull().sum()) #لا يوجد قيم مفقودة
print(orders.duplicated().sum()) # لا يوجد تكرارات
```

```python
total_customers_orders = pd.merge(customers,orders,on = "customer_id",how = "left")
print(total_customers_orders)
# هون احنا عملنا جدول فيه كل العملاء سواء طلبو او لا  ( يلي طلب اخد طلبه من جدول الطلبات و يلي ما طلب جابه و ترك القيم فارغة)
```

```python
customer_total_spent = total_customers_orders.groupby("Customer Name")["amount"].sum().sort_values(ascending = False)
print("\n مجموع طلبات كل عميل")
print(customer_total_spent)

print("\n عدد طلبات كل عميل")
customer_total_orders = total_customers_orders.groupby("Customer Name")["order_id"].count().sort_values(ascending = False)
print(customer_total_orders)

print("\n Top 3 Customers")
print(customer_total_spent.head(3))

print("\n Bottom 3 Customers")
print(customer_total_spent.tail(3))

cities_total_sales = total_customers_orders.groupby("City")["amount"].sum().sort_values(ascending = False)
print("\n أعلى 3 مدن من حيث المبيعات")
print(cities_total_sales.head(3))

print("\n أقل 3 مدن من حيث المبيعات ")
print(cities_total_sales.tail(3))
```
